```python
df = pd.read_csv("employee_birthdates.csv")

df.head()
```

```python
df["birth_date"] = pd.to_datetime(df["birth_date"])
```

```python
df["birth_year"] = df["birth_date"].dt.year

df.head()
```

```python
oldest_three = df.sort_values(by="birth_date").head(3)

oldest_three[["employee_id", "birth_date"]]
```

```python
birth_year_count = df["birth_year"].value_counts().sort_index()

print(birth_year_count)
```

```python
born_after_1990 = df[df["birth_year"] > 1990]

print(born_after_1990)
```

```python
today = pd.Timestamp.today()

df["age"] = today.year - df["birth_date"].dt.year

df.loc[
    (df["birth_date"].dt.month > today.month) |
    (
        (df["birth_date"].dt.month == today.month) &
        (df["birth_date"].dt.day > today.day)
    ),
    "age"
] -= 1

df[["employee_id", "birth_date", "age"]]
```
